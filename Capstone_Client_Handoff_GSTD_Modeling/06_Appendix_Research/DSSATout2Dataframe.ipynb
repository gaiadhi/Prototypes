```python
# Lazy process: read the OUT file, 
# remove unnecessary lines after extracting needed run and treatment data 
# to the necessary lines add run and treatment run and treatment data
# write those lines out to a temp.OUT file
# Use pd.read_csv('temp.OUT', sep="\s+") to get the df
import pandas as pd

fout = open('temp.OUT', 'w')
got_colnames = False
with open('PlantGro.OUT') as file:
    while line := file.readline():
        #print(line.rstrip())
        if line.startswith ('*RUN'):
            runno = line.split()[1]
            print (f'starting Run {runno}')
        elif line.startswith (' TREATMENT'):
            trtno = line.split()[1]
            trtmnt = line.split()[3]
            #print(f'starting trtmnt {trtno}: {trtmnt}')
        elif line.startswith(' DATA') or line.startswith('!') or len(line.rstrip()) == 0 or line.startswith('*') or line.startswith(' MODEL') or line.startswith(' EXPERIMENT'):
            continue
        else:
            #print (f'appending {line}')
            if not line.startswith('@'):
                line = line.rstrip() + f'       {runno}    {trtno}                        {trtmnt}\n'
                fout.write(line)
            elif not got_colnames:
                got_colnames = True
                line = line.rstrip() + f'    RUNNO    TRTNO                        TRTNA\n'   
                fout.write(line)
                
fout.close()
file.close()

df = pd.read_csv('temp.OUT', sep="\s+")
df.to_pickle('PlantGro.pkl')
```
