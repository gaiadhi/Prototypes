```python
import shapely
from shapely.geometry import shape, GeometryCollection
import json
```

```python
geo_str = '{"type": "FeatureCollection", "features": [{"type":"Feature","geometry":{"type":"Polygon","coordinates":[[[77.29037314618061,11.12217160979539],[77.45094275390798,11.120875293718415],[77.28020744237891,10.787298351561953],[76.76688365648427,11.04600779755218],[76.76523089289189,11.047079511259524],[76.75084766779325,11.11773216918325],[77.29087396299926,11.120469785813327],[77.29037314618061,11.12217160979539]]]},"properties":null,"id":"Coimbatore, Tamil Nadu"}]}'
```

```python
geo_json = json.loads(geo_str)
```

```python
features = geo_json['features']
```

```python
features
```

```python
geocoll = GeometryCollection([shape(feature["geometry"]).buffer(0) for feature in features])
```

```python
for geo in geocoll.geoms:
    print (f' centroid: {geo.centroid}, bounding rect: {geo.bounds}')
```
